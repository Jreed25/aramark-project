# Aramark–Avendra Spend Analysis (EDA)

## Objective
This notebook explores the Aramark–Avendra dataset to identify spending patterns across customers, industries, locations, and product categories.

We aim to:
- Understand spending distribution
- Identify key drivers of spend
- Explore geographic and business differences
- Generate insights for future modeling

## Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Load Data

We load the dataset containing customer spending data. The dataset includes information about categories, locations, business types, and total spend.

```python
import pandas as pd

chunks = []
for chunk in pd.read_csv(
    "gs://cs-562-aramark-project/Andrew_Meszaros_SRF_2026-04-01-0936.csv",
    chunksize=100000
):
    chunks.append(chunk)

df = pd.concat(chunks, ignore_index=True)
print(df.shape)
print(df.head())
```

## Dataset Overview
We first inspect the structure of the dataset including shape, data types, and summary statistics.

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

## Data Cleaning
We check for missing values and handle basic cleaning before analysis.

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Basic cleaning (adjust if needed)
df = df.dropna(subset=["Spend Random Factor"])
```

## Exploratory Analysis


## Spend by Category Level 1
We analyze which product categories contribute the most to total spend.

```python
category_spend = df.groupby("Category Level 1")["Spend Random Factor"].sum().sort_values(ascending=False)
category_spend
```

```python
category_spend.plot(kind="bar", figsize=(10,5))
plt.title("Total Spend by Category Level 1")
plt.ylabel("Spend")
plt.show()
```

## Spend by State
We analyze geographic distribution of spend across US states.

```python
state_spend = df.groupby("State")["Spend Random Factor"].sum().sort_values(ascending=False)
state_spend.head(10)
```

```python
state_spend.head(10).plot(kind="bar", figsize=(10,5))
plt.title("Top 10 States by Spend")
plt.show()
```

## Spend by Business Entity Type
We compare spending behavior across different industries.

```python
business_spend = df.groupby("Business Entity Type")["Spend Random Factor"].mean().sort_values(ascending=False)
business_spend
```

```python
business_spend.plot(kind="bar", figsize=(10,5))
plt.title("Average Spend by Business Type")
plt.show()
```

## Spend by City
We examine high-spending cities in the dataset.

```python
city_spend = df.groupby("City")["Spend Random Factor"].sum().sort_values(ascending=False)
city_spend.head(10)
```

## Ecommerce vs Non-Ecommerce Spending
We analyze whether ecommerce adoption impacts spending behavior.

```python
ecom_spend = df.groupby("Ecommerce Status")["Spend Random Factor"].mean()
ecom_spend
```

```python
ecom_spend.plot(kind="bar")
plt.title("Ecommerce vs Non-Ecommerce Spend")
plt.show()
```

## Key Insights

- Spending is concentrated in a few major categories
- Geographic distribution shows strong regional patterns
- Business type significantly impacts spending behavior
- Ecommerce usage may influence customer spending

These insights provide a strong foundation for further analysis such as clustering or predictive modeling.
